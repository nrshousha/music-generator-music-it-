```python
import pandas as pd
import pretty_midi

# 1. look at the metadata
df = pd.read_csv('../data/midi/maestro-v3.0.0/maestro-v3.0.0.csv')
print(df.shape)           # how many files
print(df.head())          # columns: year, composer, title, split, duration, midi_filename
print(df['split'].value_counts())   # train/validation/test split already done for you
```

```python
# 2. open one MIDI file and inspect it
midi_path = '../data/midi/maestro-v3.0.0/' + df['midi_filename'].iloc[0]
midi = pretty_midi.PrettyMIDI(midi_path)

print(f"Duration: {midi.get_end_time():.1f} seconds")
print(f"Instruments: {[i.name for i in midi.instruments]}")

notes = midi.instruments[0].notes
print(f"Total notes: {len(notes)}")
print(f"First note → pitch: {notes[0].pitch}, start: {notes[0].start:.2f}s, velocity: {notes[0].velocity}")
```
